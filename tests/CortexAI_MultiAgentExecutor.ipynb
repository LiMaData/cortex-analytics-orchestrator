## Data Analyst Agent

````python

class CortexAnalyst:
    """Cortex Analyst simulator using CORTEX.COMPLETE"""
    
    def __init__(self, session, semantic_model_stage):
        self.session = session
        self.semantic_model = self._load_semantic_model(semantic_model_stage)
        
        # IMPROVED: Use strategic context instead of first 2500 chars
        self.schema_context = self._build_context()
        
        print(f"✅ Loaded semantic model ({len(self.semantic_model)} chars)")
        print(f"✅ Built context ({len(self.schema_context)} chars)")
        
    def _load_semantic_model(self, stage_path):
        """Load semantic model from Snowflake stage"""
        # Create file format
        self.session.sql("""
            CREATE FILE FORMAT IF NOT EXISTS yaml_format
            TYPE = 'CSV'
            FIELD_DELIMITER = NONE
            RECORD_DELIMITER = NONE
        """).collect()
        
        # Read file
        result = self.session.sql(f"""
            SELECT t.$1 AS content
            FROM {stage_path} (FILE_FORMAT => yaml_format) t
        """).collect()
        
        return "\n".join([row['CONTENT'] for row in result])
    
    def _build_context(self):
        """Build strategic context for better generalization"""
        
        # Get critical sections
        header = self.semantic_model[:800]  # Name, physical columns
        
        # Extract verified queries
        verified_start = self.semantic_model.find('verified_queries:')
        verified_end = self.semantic_model.find('tables:', verified_start)
        verified = self.semantic_model[verified_start:verified_end] if verified_start != -1 else ""
        
        # Get first table definition
        tables_start = self.semantic_model.find('tables:')
        first_table = self.semantic_model[tables_start:tables_start+2000] if tables_start != -1 else ""
        
        return f"{header}\n\n{verified}\n\n{first_table}"
    
    def _clean_sql(self, text):
        """Remove markdown and formatting from SQL"""
        text = re.sub(r'```sql\s*', '', text, flags=re.IGNORECASE)
        text = re.sub(r'```\s*', '', text)
        text = text.strip()
        
        # Check for explanations
        explanation_keywords = ['to answer', 'this query', 'here is', 'we need', 'first']
        if any(text.lower().startswith(keyword) for keyword in explanation_keywords):
            # Try to extract SQL
            match = re.search(r'((?:WITH|SELECT).*?)(?:;|\Z)', text, re.IGNORECASE | re.DOTALL)
            if match:
                text = match.group(1)
            else:
                raise ValueError("LLM returned explanation without SQL")
        
        return text.rstrip(';').strip()
    
    def ask(self, question):
        # Simplified prompt
        prompt = f"""Generate SQL query for Snowflake.

    Schema info:
    {self.schema_context[:1000]}

    Rules:
    - Return ONLY the SQL query
    - No explanations
    - No markdown
    - No extra text

    Question: {question}

    SQL:"""
        
        result = self.session.sql(f"""
            SELECT SNOWFLAKE.CORTEX.COMPLETE('mistral-large', '{prompt.replace("'", "''")}') AS sql_text
        """).collect()
        
        raw_sql = result[0]['SQL_TEXT']
        
        # Debug: print what we got
        print(f"   [DEBUG] Raw LLM output: {raw_sql[:200]}...")
        
        clean_sql = self._clean_sql(raw_sql)
        
        print(f"   [DEBUG] Clean SQL: {clean_sql[:200]}...")
        
        try:
            data = self.session.sql(clean_sql).collect()
            results = []
            for row in data:
                row_dict = row.asDict()
                converted_row = {}
                for key, value in row_dict.items():
                    if isinstance(value, str):
                        try:
                            if '.' in value:
                                converted_row[key] = float(value)
                            else:
                                converted_row[key] = int(value)
                        except (ValueError, AttributeError):
                            converted_row[key] = value
                    else:
                        converted_row[key] = value
                results.append(converted_row)
            return {'success': True, 'sql': clean_sql, 'results': results, 'row_count': len(results)}
        except Exception as e:
            return {'success': False, 'sql': clean_sql, 'results': [], 'error': str(e)}

print("✅ CortexAnalyst class defined")
````

## Data Visualization Agent

````python
import plotly.graph_objects as go
import plotly.express as px

class VisualizationAgent:
    """Agent for creating data visualizations"""
    
    def __init__(self):
        print("✅ VisualizationAgent initialized")
    
    def create_visualization(self, data, question=None, chart_type=None):
        """Auto-create appropriate visualization"""
        if not data:
            print("⚠️ No data to visualize")
            return None
        
        df = pd.DataFrame(data)
        
        # Auto-convert numeric columns
        for col in df.columns:
            if col.upper() not in ['BUSINESSUNIT', 'MARKET', 'COUNTRY_CODE', 
                                   'CHANNEL_GROUPING', 'SOURCE', 'CAMPAIGN']:
                try:
                    df[col] = pd.to_numeric(df[col])
                except:
                    pass
        
        # Smart categorical detection
        if chart_type is None:
            first_col = df.columns[0].upper()
            categorical_business_dims = ['BUSINESSUNIT', 'MARKET', 'COUNTRY_CODE', 
                                        'COUNTRY', 'CHANNEL', 'CHANNEL_GROUPING', 
                                        'REGION', 'SEGMENT', 'CAMPAIGN']
            
            if first_col in categorical_business_dims:
                print(f"   Detected categorical dimension: {first_col}")
                chart_type = 'bar'
            else:
                chart_type = self._infer_chart_type(df, question)
        
        print(f"📊 Creating {chart_type} chart...")
        
        if chart_type == 'bar':
            return self._create_bar_chart(df, question)
        elif chart_type == 'line':
            return self._create_line_chart(df, question)
        elif chart_type == 'pie':
            return self._create_pie_chart(df, question)
        else:
            return self._create_bar_chart(df, question)
    
    def _infer_chart_type(self, df, question=None):
        """Infer chart type"""
        
        if question:
            question_lower = question.lower()
            
            if 'trend' in question_lower or 'over time' in question_lower:
                return 'line'
            elif 'by market' in question_lower or 'by country' in question_lower:
                return 'bar'
        
        # Check for date columns
        for col in df.columns:
            if 'date' in col.lower() or 'time' in col.lower():
                return 'line'
        
        return 'bar'  # Default
    
    def _create_bar_chart(self, df, title=None):
        """Create bar chart"""
        
        df.columns = [col.upper() for col in df.columns]
        
        numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
        string_cols = df.select_dtypes(include=['object', 'string']).columns.tolist()
        
        if not numeric_cols:
            print("⚠️ No numeric columns")
            return None
        
        x_col = string_cols[0] if string_cols else df.columns[0]
        y_col = numeric_cols[0]
        
        # Sort by value
        df_sorted = df.sort_values(by=y_col, ascending=False)
        
        fig = px.bar(
            df_sorted,
            x=x_col,
            y=y_col,
            title=title or f"{y_col} by {x_col}",
            color=x_col,
            text=y_col,
            color_discrete_sequence=px.colors.qualitative.Set2
        )
        
        # Format numbers
        max_val = df[y_col].max()
        if max_val > 1000000:
            text_template = '%{text:,.0f}'
        else:
            text_template = '%{text:,.2f}'
        
        fig.update_traces(
            texttemplate=text_template,
            textposition='outside'
        )
        
        fig.update_layout(
            xaxis_title=x_col.replace('_', ' ').title(),
            yaxis_title=y_col.replace('_', ' ').title(),
            showlegend=False,
            height=600
        )
        
        return fig
    
    def _create_line_chart(self, df, title=None):
        """Create line chart"""
        x_col = df.columns[0]
        numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
        
        fig = go.Figure()
        for y_col in numeric_cols:
            fig.add_trace(go.Scatter(
                x=df[x_col], y=df[y_col],
                mode='lines+markers',
                name=y_col
            ))
        
        fig.update_layout(title=title or "Trend", height=600)
        return fig
    
    def _create_pie_chart(self, df, title=None):
        """Create pie chart"""
        cat_cols = df.select_dtypes(exclude=['number']).columns.tolist()
        num_cols = df.select_dtypes(include=['number']).columns.tolist()
        
        labels = cat_cols[0] if cat_cols else df.columns[0]
        values = num_cols[0] if num_cols else df.columns[1]
        
        fig = px.pie(df, names=labels, values=values, title=title)
        fig.update_layout(height=600)
        return fig

print("✅ VisualizationAgent class defined")
````

## Insight Agent

````python
class InsightAgent:
    """Generate narrative insights using LLM"""
    
    def __init__(self, session):
        self.session = session
        print("✅ InsightAgent initialized")
    
    def generate_insights(self, data, question):
        """Generate natural language insights"""
        if not data or len(data) == 0:
            return "No data available for analysis."
        
        # Prepare data summary
        df = pd.DataFrame(data)
        summary = []
        
        # Top 3 rows
        top_rows = df.head(3).to_string(index=False)
        summary.append(f"Top Results:\n{top_rows}")
        
        # Basic stats
        numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
        if numeric_cols:
            for col in numeric_cols[:2]:
                summary.append(f"{col}: Total={df[col].sum():,.0f}, Avg={df[col].mean():,.1f}")
        
        data_summary = "\n".join(summary)
        
        # Generate insights with LLM
        prompt = f"""You are a business analyst. Provide 2-3 key insights in bullet points.

QUESTION: {question}

DATA:
{data_summary}

KEY INSIGHTS (2-3 bullets):"""

        try:
            result = self.session.sql(f"""
                SELECT SNOWFLAKE.CORTEX.COMPLETE(
                    'mistral-large',
                    '{prompt.replace("'", "''")}'
                ) AS insights
            """).collect()
            
            return result[0]['INSIGHTS']
        except:
            return " | ".join(summary)

print("✅ InsightAgent class defined")
````

## MultiAgents Executor

````python
class MultiAgentExecutor:
    """Orchestrator that manages multiple agents and workflow"""
    
    def __init__(self, session, semantic_model_stage):
        print("=" * 60)
        print("🎯 INITIALIZING MULTI-AGENT SYSTEM")
        print("=" * 60)
        
        self.session = session
        
        # Initialize specialized agents
        print("\n📦 Loading Agents:")
        self.analyst = CortexAnalyst(session, semantic_model_stage)
        self.viz_agent = VisualizationAgent()
        
        print("\n✅ SYSTEM READY")
        print("   • Data Analyst Agent: CortexAnalyst")
        print("   • Visualization Agent: VisualizationAgent")
        print("=" * 60)
    
    def execute(self, question, options=None):
        """
        Main orchestration method
        Decides workflow and coordinates agents
        """
        options = options or {}
        
        print("\n" + "=" * 60)
        print(f"🤖 EXECUTOR: New Query Received")
        print("=" * 60)
        print(f"❓ Question: {question}")
        
        # Step 1: Query Classification
        print("\n📋 STEP 1: Query Classification")
        query_type = self._classify_query(question)
        print(f"   └─ Type: {query_type}")
        
        # Step 2: Route to Data Analyst Agent
        print("\n📊 STEP 2: Routing to Data Analyst Agent")
        print("   └─ CortexAnalyst generating SQL...")
        data_result = self.analyst.ask(question)
        
        if not data_result['success']:
            print(f"   └─ ❌ Error: {data_result.get('error')}")
            return {
                'success': False,
                'error': data_result.get('error'),
                'sql': data_result.get('sql')
            }
        
        print(f"   └─ ✅ Data retrieved: {data_result['row_count']} rows")
        print(f"   └─ SQL: {data_result['sql'][:80]}...")
        
        # Step 3: Route to Visualization Agent
        needs_viz = options.get('visualize', True)
        chart = None
        
        if needs_viz and data_result['row_count'] > 0:
            print("\n📈 STEP 3: Routing to Visualization Agent")
            print("   └─ VisualizationAgent creating chart...")
            chart = self.viz_agent.create_visualization(
                data_result['results'], 
                question=question,
                chart_type=options.get('chart_type')
            )
            print("   └─ ✅ Visualization created")
        else:
            print("\n⊘ STEP 3: Visualization skipped")
        
        # Step 4: Generate Insights
        print("\n💡 STEP 4: Generating Insights")
        insights = self._generate_insights(data_result['results'], question)
        print(f"   └─ {insights}")
        
        print("\n" + "=" * 60)
        print("✅ WORKFLOW COMPLETE")
        print("=" * 60)
        
        # Return the complete result
        return {
            'success': True,
            'query': question,
            'query_type': query_type,
            'sql': data_result['sql'],
            'data': data_result['results'],
            'visualization': chart,
            'insights': insights,
            'row_count': data_result['row_count']
        }
    
    def _classify_query(self, question):
        """Classify query type to determine workflow"""
        question_lower = question.lower()
        
        # Pattern matching for query classification
        if any(word in question_lower for word in ['trend', 'over time', 'weekly', 'monthly']):
            return '📈 TREND_ANALYSIS'
        elif any(word in question_lower for word in ['by market', 'by country', 'by channel']):
            return '📊 COMPARISON'
        elif any(word in question_lower for word in ['total', 'sum', 'count']):
            return '📋 AGGREGATION'
        elif any(word in question_lower for word in ['top', 'bottom', 'best', 'worst']):
            return '🏆 RANKING'
        else:
            return '📋 GENERAL'
    
    def _generate_insights(self, data, question):
        """Generate natural language insights from data"""
        if not data:
            return "No data available for analysis."
        
        df = pd.DataFrame(data)
        
        # Basic insights
        insights = []
        insights.append(f"Found {len(data)} records")
        
        # Find numeric columns
        numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
        
        if numeric_cols:
            for col in numeric_cols[:2]:  # Top 2 metrics
                total = df[col].sum()
                avg = df[col].mean()
                insights.append(f"{col} Total: {total:,.0f}, Avg: {avg:,.1f}")
        
        return " | ".join(insights)

print("✅ MultiAgentExecutor class defined")
````

````python
## LLM-based classifier for smarter routing: Query Intent Classification (Enhanced Routing)
def _classify_query_with_llm(self, question: str) -> dict:
    """Use LLM to understand query intent and required tools"""
    
    prompt = f"""Analyze this business question and classify it:

QUESTION: {question}

Return JSON with:
{{
    "query_type": "TREND|COMPARISON|AGGREGATION|RANKING",
    "needs_sql": true/false,
    "needs_visualization": true/false,
    "suggested_chart": "bar|line|pie|table",
    "complexity": "simple|medium|complex"
}}

JSON:"""

    result = self.session.sql(f"""
        SELECT SNOWFLAKE.CORTEX.COMPLETE(
            'mistral-large',
            '{prompt.replace("'", "''")}'
        ) AS classification
    """).collect()
    
    import json
    return json.loads(result[0]['CLASSIFICATION'])
````

````python
class MultiAgentExecutor:
    """Orchestrator that manages multiple agents and workflow"""
    
    def __init__(self, session, semantic_model_stage):
        print("=" * 60)
        print("🎯 INITIALIZING MULTI-AGENT SYSTEM")
        print("=" * 60)
        
        self.session = session
        
        # Initialize specialized agents
        print("\n📦 Loading Agents:")
        self.analyst = CortexAnalyst(session, semantic_model_stage)
        self.viz_agent = VisualizationAgent()
        
        print("\n✅ SYSTEM READY")
        print("   • Data Analyst Agent: CortexAnalyst")
        print("   • Visualization Agent: VisualizationAgent")
        print("=" * 60)
    
    def execute(self, question, options=None):
        """Main orchestration method - Decides workflow and coordinates agents"""
        options = options or {}
        
        print("\n" + "=" * 60)
        print(f"🤖 EXECUTOR: New Query Received")
        print("=" * 60)
        print(f"❓ Question: {question}")
        
        # Step 1: Query Classification
        print("\n📋 STEP 1: Query Classification")
        query_type = self._classify_query(question)
        print(f"   └─ Type: {query_type}")
        
        # Step 2: Route to Data Analyst Agent
        print("\n📊 STEP 2: Routing to Data Analyst Agent")
        print("   └─ CortexAnalyst generating SQL...")
        data_result = self.analyst.ask(question)
        
        if not data_result['success']:
            print(f"   └─ ❌ Error: {data_result.get('error')}")
            return {
                'success': False,
                'error': data_result.get('error'),
                'sql': data_result.get('sql')
            }
        
        print(f"   └─ ✅ Data retrieved: {data_result['row_count']} rows")
        print(f"   └─ SQL: {data_result['sql'][:80]}...")
        
        # Step 3: Route to Visualization Agent
        needs_viz = options.get('visualize', True)
        chart = None
        
        if needs_viz and data_result['row_count'] > 0:
            print("\n📈 STEP 3: Routing to Visualization Agent")
            print("   └─ VisualizationAgent creating chart...")
            chart = self.viz_agent.create_visualization(
                data_result['results'], 
                question=question,
                chart_type=options.get('chart_type')
            )
            print("   └─ ✅ Visualization created")
        else:
            print("\n⊘ STEP 3: Visualization skipped")
        
        # Step 4: Generate Insights
        print("\n💡 STEP 4: Generating Insights")
        insights = self._generate_insights(data_result['results'], question)
        print(f"   └─ {insights}")
        
        print("\n" + "=" * 60)
        print("✅ WORKFLOW COMPLETE")
        print("=" * 60)
        
        # Return the complete result
        return {
            'success': True,
            'query': question,
            'query_type': query_type,
            'sql': data_result['sql'],
            'data': data_result['results'],
            'visualization': chart,
            'insights': insights,
            'row_count': data_result['row_count']
        }
    
    def _classify_query(self, question):
        """Classify query type to determine workflow"""
        question_lower = question.lower()
        
        if any(word in question_lower for word in ['trend', 'over time', 'weekly', 'monthly']):
            return '📈 TREND_ANALYSIS'
        elif any(word in question_lower for word in ['by market', 'by country', 'by channel']):
            return '📊 COMPARISON'
        elif any(word in question_lower for word in ['total', 'sum', 'count']):
            return '📋 AGGREGATION'
        elif any(word in question_lower for word in ['top', 'bottom', 'best', 'worst']):
            return '🏆 RANKING'
        else:
            return '📋 GENERAL'
    
    def _generate_insights(self, data, question):
        """Generate natural language insights from data"""
        if not data:
            return "No data available for analysis."
        
        df = pd.DataFrame(data)
        insights = []
        insights.append(f"Found {len(data)} records")
        
        numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
        if numeric_cols:
            for col in numeric_cols[:2]:
                total = df[col].sum()
                avg = df[col].mean()
                insights.append(f"{col} Total: {total:,.0f}, Avg: {avg:,.1f}")
        
        return " | ".join(insights)

print("✅ MultiAgentExecutor class defined (CLEAN VERSION)")
````

````python
# ===== COMPLETE SYSTEM RESTART =====

from snowflake.snowpark import Session
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import re

print("🔄 RESTARTING ENTIRE SYSTEM...\n")

# 1. RECONNECT TO SNOWFLAKE
print("1️⃣ Reconnecting to Snowflake...")

connection_parameters = {
    "account": "wkswaox-lt08934",
    "user": "Lima717",
    "password": "Easy2snowflake!",
    "role": "ACCOUNTADMIN",
    "warehouse": "COMPUTE_WH",
    "database": "CAMPAIGN_ANALYTICS",
    "schema": "GENERATED_DATA"
}

session = Session.builder.configs(connection_parameters).create()
print("   ✅ Connected!\n")

# 2. RELOAD CORTEX ANALYST
print("2️⃣ Loading CortexAnalyst...")

class CortexAnalyst:
    def __init__(self, session, semantic_model_stage):
        self.session = session
        self.semantic_model = self._load_semantic_model(semantic_model_stage)
        self.schema_context = self._build_context()
        
    def _load_semantic_model(self, stage_path):
        self.session.sql("""
            CREATE FILE FORMAT IF NOT EXISTS yaml_format
            TYPE = 'CSV' FIELD_DELIMITER = NONE RECORD_DELIMITER = NONE
        """).collect()
        result = self.session.sql(f"""
            SELECT t.$1 AS content FROM {stage_path} (FILE_FORMAT => yaml_format) t
        """).collect()
        return "\n".join([row['CONTENT'] for row in result])
    
    def _build_context(self):
        header = self.semantic_model[:800]
        verified_start = self.semantic_model.find('verified_queries:')
        verified_end = self.semantic_model.find('tables:', verified_start)
        verified = self.semantic_model[verified_start:verified_end] if verified_start != -1 else ""
        tables_start = self.semantic_model.find('tables:')
        first_table = self.semantic_model[tables_start:tables_start+2000] if tables_start != -1 else ""
        return f"{header}\n\n{verified}\n\n{first_table}"
    
    def _clean_sql(self, text):
        text = re.sub(r'```sql\s*', '', text, flags=re.IGNORECASE)
        text = re.sub(r'```\s*', '', text)
        text = text.strip()
        explanation_keywords = ['to answer', 'this query', 'here is', 'we need', 'first']
        if any(text.lower().startswith(keyword) for keyword in explanation_keywords):
            match = re.search(r'((?:WITH|SELECT).*?)(?:;|\Z)', text, re.IGNORECASE | re.DOTALL)
            if match:
                text = match.group(1)
            else:
                raise ValueError("LLM returned explanation without SQL")
        return text.rstrip(';').strip()
    
    def ask(self, question):
        prompt = f"""You are a SQL generator. Learn from VERIFIED QUERIES examples and apply patterns.

{self.schema_context}

PATTERN RULES:
1. "X by market" -> GROUP BY BUSINESSUNIT or COUNTRY_CODE
2. "X this week" -> WHERE date >= DATE_TRUNC('week', CURRENT_DATE())
3. "X last month" -> WHERE date >= DATEADD(month, -1, CURRENT_DATE())
4. "rate" -> (SUM(numerator) / NULLIF(SUM(denominator), 0)) * 100
5. Specific market -> WHERE BUSINESSUNIT = 'VCUS' or COUNTRY_CODE = 'VCUS'

CRITICAL: Return ONLY SQL code, NO explanations.

QUESTION: {question}

SQL CODE:"""
        result = self.session.sql(f"""
            SELECT SNOWFLAKE.CORTEX.COMPLETE('mistral-large', '{prompt.replace("'", "''")}') AS sql_text
        """).collect()
        raw_sql = result[0]['SQL_TEXT']
        clean_sql = self._clean_sql(raw_sql)
        
        try:
            data = self.session.sql(clean_sql).collect()
            results = []
            for row in data:
                row_dict = row.asDict()
                converted_row = {}
                for key, value in row_dict.items():
                    if isinstance(value, str):
                        try:
                            if '.' in value:
                                converted_row[key] = float(value)
                            else:
                                converted_row[key] = int(value)
                        except (ValueError, AttributeError):
                            converted_row[key] = value
                    else:
                        converted_row[key] = value
                results.append(converted_row)
            return {'success': True, 'sql': clean_sql, 'results': results, 'row_count': len(results)}
        except Exception as e:
            return {'success': False, 'sql': clean_sql, 'results': [], 'error': str(e)}

print("   ✅ CortexAnalyst loaded\n")

# 3. RELOAD VISUALIZATION AGENT
print("3️⃣ Loading VisualizationAgent...")

class VisualizationAgent:
    def __init__(self):
        pass
    
    def create_visualization(self, data, question=None, chart_type=None):
        if not data:
            return None
        df = pd.DataFrame(data)
        for col in df.columns:
            if col.upper() not in ['BUSINESSUNIT', 'MARKET', 'COUNTRY_CODE', 'CHANNEL_GROUPING', 'SOURCE', 'CAMPAIGN']:
                try:
                    df[col] = pd.to_numeric(df[col])
                except:
                    pass
        
        if chart_type is None:
            first_col = df.columns[0].upper()
            categorical_dims = ['BUSINESSUNIT', 'MARKET', 'COUNTRY_CODE', 'COUNTRY', 'CHANNEL', 'CHANNEL_GROUPING']
            chart_type = 'bar' if first_col in categorical_dims else 'bar'
        
        return self._create_bar_chart(df, question) if chart_type == 'bar' else None
    
    def _create_bar_chart(self, df, title=None):
        df.columns = [col.upper() for col in df.columns]
        numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
        string_cols = df.select_dtypes(include=['object', 'string']).columns.tolist()
        
        if not numeric_cols:
            return None
        
        x_col = string_cols[0] if string_cols else df.columns[0]
        y_col = numeric_cols[0]
        df_sorted = df.sort_values(by=y_col, ascending=False)
        
        fig = px.bar(df_sorted, x=x_col, y=y_col, title=title or f"{y_col} by {x_col}",
                     color=x_col, text=y_col, color_discrete_sequence=px.colors.qualitative.Set2)
        fig.update_traces(texttemplate='%{text:,.0f}', textposition='outside')
        fig.update_layout(xaxis_title=x_col.replace('_', ' ').title(),
                         yaxis_title=y_col.replace('_', ' ').title(),
                         showlegend=False, height=600)
        return fig

print("   ✅ VisualizationAgent loaded\n")

# 4. RELOAD EXECUTOR
print("4️⃣ Loading MultiAgentExecutor...")

class MultiAgentExecutor:
    def __init__(self, session, semantic_model_stage):
        self.session = session
        self.analyst = CortexAnalyst(session, semantic_model_stage)
        self.viz_agent = VisualizationAgent()
    
    def execute(self, question, options=None):
        options = options or {}
        print(f"\n🤖 Processing: {question}")
        
        query_type = self._classify_query(question)
        print(f"   Type: {query_type}")
        
        data_result = self.analyst.ask(question)
        if not data_result['success']:
            return {'success': False, 'error': data_result.get('error'), 'sql': data_result.get('sql')}
        
        print(f"   ✅ {data_result['row_count']} rows")
        
        chart = None
        if options.get('visualize', True) and data_result['row_count'] > 0:
            chart = self.viz_agent.create_visualization(data_result['results'], question)
        
        insights = self._generate_insights(data_result['results'], question)
        
        return {
            'success': True, 'query': question, 'query_type': query_type,
            'sql': data_result['sql'], 'data': data_result['results'],
            'visualization': chart, 'insights': insights, 'row_count': data_result['row_count']
        }
    
    def _classify_query(self, question):
        q = question.lower()
        if 'trend' in q or 'over time' in q:
            return 'TREND'
        elif 'by market' in q or 'by country' in q:
            return 'COMPARISON'
        elif 'top' in q or 'bottom' in q:
            return 'RANKING'
        else:
            return 'GENERAL'
    
    def _generate_insights(self, data, question):
        if not data:
            return "No data"
        df = pd.DataFrame(data)
        insights = [f"{len(data)} records"]
        numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
        if numeric_cols:
            col = numeric_cols[0]
            insights.append(f"{col}: {df[col].sum():,.0f}")
        return " | ".join(insights)

print("   ✅ MultiAgentExecutor loaded\n")

# 5. INITIALIZE EXECUTOR
print("5️⃣ Initializing system...")

executor = MultiAgentExecutor(session, '@semantic_models/marketing_semantic_model.yaml')  # UPDATE PATH

print("   ✅ System initialized\n")

# 6. TEST
print("6️⃣ Testing...")

test = executor.execute("Total sends by market", options={'visualize': False})

if test['success']:
    print(f"   ✅ TEST PASSED - {test['row_count']} rows\n")
    print("=" * 60)
    print("🎉 SYSTEM READY FOR DEMO!")
    print("=" * 60)
else:
    print(f"   ❌ TEST FAILED: {test.get('error')}\n")
````
